```python
#here P is planar and G is the complement of P
def n_choose_2(n):
    return n*(n-1)/2

def poly1(n): #lower bound on e(G^s)
    return n**3/4 - 11/4*n**2 + 13.5*n-33 + 36/n

def poly2(n):#upper bound on e(P^s)
    s = (9*n**2 - 36*n + 36)/(n-1) + 2*n**2 - 17/2*n + 9
    # print(abs(t-s) < 1e-9)
    return s


for n in range(10,50):
    if poly1(n)>poly2(n):
        print(f"n is {n}, no minimal counterexample on {n} vertices")
```

```python
def f(n,eG): #lower bound on the edges of G^s
    return eG**2/n + 3/2*eG + n/2

def g(n,eP,nb_triangles):#upper bound on the edges of P^s
    C_4bound = 1/2*(n**2-5*n +6)
    K12_bound = eP**2/(n-1) + eP*(n-2)/2
    # print(abs(t-s) < 1e-9)
    return C_4bound + K12_bound -3*nb_triangles
#here P is planar and G is the complement of P
N=13
for eP in range(28,33):
    eG = n_choose_2(N)-eP
    # print(f"eG is {eG}")
    if f(N,eG)>g(N,eP,1):
        print(f"No minimal counterexample on {N} vertices and {eG} edges, edges of the graph P = {eP}")

N=15
for eP in range(36,39):
    eG = n_choose_2(N)-eP
    # print(f"eG is {eG}")
    if f(N,eG)>g(N,eP,1):
        print(f"No minimal counterexample on {N} vertices and {eG} edges, edges of the graph P = {eP}")
```
